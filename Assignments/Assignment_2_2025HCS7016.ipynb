**Question 2.** A ball is thrown straight up at u = 20 m/s. Its height is h = u t − ½ g t². Rearranged, (g/2) t² − u t + h = 0.

For a target height h, compute the discriminant and print one of three messages: the ball reaches h twice (print both times), just touches h at the top, or never reaches h (print the maximum height u²/2g).

Test with h = 15 m and h = 25 m.

```python
# Question 2
import math

u = 20.0
g = 9.81

for h in [15.0, 25.0]:
    discriminant = u**2 - 4 * (g / 2) * h
    print(f"Target height = {h} m")

    if discriminant > 0:
        t1 = (u - math.sqrt(discriminant)) / g
        t2 = (u + math.sqrt(discriminant)) / g
        print(f"The ball reaches this height twice: {t1:.3f} s and {t2:.3f} s")
    elif discriminant == 0:
        t = u / g
        print(f"The ball just touches this height at the top: {t:.3f} s")
    else:
        maximum_height = u**2 / (2 * g)
        print(f"The ball never reaches this height. Maximum height = {maximum_height:.3f} m")
```

```python
# Question 10
import math

def proj_range(v0, theta_deg=45.0, g=9.81):
    theta = math.radians(theta_deg)
    return v0**2 * math.sin(2 * theta) / g

print(f"Earth, 45 degrees: {proj_range(20):.2f} m")
print(f"Earth, 30 degrees: {proj_range(20, 30):.2f} m")
print(f"Moon, 45 degrees: {proj_range(20, 45, 1.62):.2f} m")
print(f"Mars, 60 degrees: {proj_range(20, 60, 3.71):.2f} m")
```

**Question 13.** Write factorial(n) using a for loop. Then write exp_series(x, nterms=10) that adds the first nterms terms of Σ xⁿ / n!, calling factorial inside it.

For x = 2, print the approximation and its error compared with math.exp(2) for nterms = 2, 4, 8 and 16.

```python
# Question 13
import math

def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

def exp_series(x, nterms=10):
    total = 0.0
    for n in range(nterms):
        total += x**n / factorial(n)
    return total

for nterms in [2, 4, 8, 16]:
    approximation = exp_series(2, nterms)
    error = abs(math.exp(2) - approximation)
    print(f"nterms = {nterms:2d}: approximation = {approximation:.10f}, error = {error:.10f}")
```

**Question 14.** Newton's method for √a repeats  x → ½ (x + a/x)  starting from x = a.

Write newton_sqrt(a, tol=1e-10) that repeats this while |x² − a| > tol and returns both the estimate and the number of iterations.

Test with a = 2, 9 and 10⁶, and then with a = 2 and tol=1e-3.

```python
# Question 14
def newton_sqrt(a, tol=1e-10):
    x = a
    iterations = 0

    while abs(x**2 - a) > tol:
        x = 0.5 * (x + a / x)
        iterations += 1

    return x, iterations

for a in [2, 9, 10**6]:
    estimate, iterations = newton_sqrt(a)
    print(f"a = {a}: square root = {estimate:.10f}, iterations = {iterations}")

estimate, iterations = newton_sqrt(2, tol=1e-3)
print(f"a = 2, tol = 1e-3: square root = {estimate:.10f}, iterations = {iterations}")
```

**Question 15.** Write celsius_to_kelvin(T_C). Then write convert_all(temps) that builds and returns a new list of Kelvin values by calling celsius_to_kelvin on each element.

Use it on [−196, 0, 25, 100] (liquid nitrogen, ice, room temperature, boiling water) and print a two-column table.

```python
# Question 15
def celsius_to_kelvin(T_C):
    return T_C + 273.15

def convert_all(temps):
    kelvin_values = []
    for temp in temps:
        kelvin_values.append(celsius_to_kelvin(temp))
    return kelvin_values

celsius_values = [-196, 0, 25, 100]
kelvin_values = convert_all(celsius_values)

print(f"{'Temperature (C)':>18} | {'Temperature (K)':>18}")
for c, k in zip(celsius_values, kelvin_values):
    print(f"{c:18.2f} | {k:18.2f}")
```

**Question 20.** The trapezoidal rule approximates ∫ y dx by Σ dx (yᵢ + yᵢ₊₁)/2. With arrays, y[:-1] holds every left point and y[1:] every right point.

(a) Use 101 points to integrate sin x from 0 to π (exact answer 2).

(b) Find the work done by a spring force F = kx with k = 200 N/m from x = 0 to 0.1 m, and compare with ½ k x².

```python
# Question 20
import numpy as np

# (a) Trapezoidal approximation for the integral of sin(x) from 0 to pi.
x = np.linspace(0, np.pi, 101)
y = np.sin(x)
dx = x[1] - x[0]
integral = np.sum(dx * (y[:-1] + y[1:]) / 2)
print(f"Integral of sin(x) from 0 to pi = {integral:.8f}")
print("Exact answer = 2")

# (b) Work done by F = kx from x = 0 to 0.1 m.
k = 200.0
x_spring = np.linspace(0, 0.1, 101)
force = k * x_spring
dx_spring = x_spring[1] - x_spring[0]
work = np.sum(dx_spring * (force[:-1] + force[1:]) / 2)
exact_work = 0.5 * k * 0.1**2
print(f"Spring work (trapezoidal) = {work:.8f} J")
print(f"Spring work (exact) = {exact_work:.8f} J")
```
