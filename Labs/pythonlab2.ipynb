1. Ask for a number and print whether it is positive, negative or zero.

```python
num = int(input("Type number: "))

if num > 0:
    print("Number is positive")
elif num == 0:
    print("Number is zero")
else:
    print("Number is negative")
```

2. Ask for marks (0 to 100) and print the grade: A for 90 and above, B for 75 and above, C for 60
and above, otherwise Fail.

```python
marks = int(input("Enter Marks: "))

if marks >= 90:
    print("Grade A")
elif 90 > marks >= 75:
    print ("Grade B")
elif 75 > marks >= 60:
    print("Grade C")
else:
    print("Fail")
```

3. Ask for three numbers and print the larger one without using max().

```python
num1 = int(input("Enter 1st number: "))
num2 = int(input("Enter 2nd number: "))
num3 = int(input("Enter 3rd number: "))

if num1 > num2 and num1 > num3:
    print(num1)
elif num2 > num1 and num2 > num3:
    print(num2)
else:
    print(num3)
```

4. Ask for a single character and print whether it is a vowel or a consonant. Use lists and in.

```python
letter = input("Enter letter: ")

vowels = ['a', 'e', 'i', 'o', 'u']

if letter.lower() in vowels:
    print("Vowel")
else:
    print("Consonant")
```

5. Ask for a username and password. Print Access granted only if the username is admin and the
password is 1234.

```python
usrnm = input("Enter username: ")
pwd = int(input("Enter password"))

if usrnm == 'admin' and pwd == 1234:
    print("Access Granted")
else:
    print("Access Denied")
```

6. Ask for some text and print You typed nothing if it is empty. Use not.

```python
txt = input("Enter your text: ")

if not txt:
    print("You typed nothing")
else:
    print(txt)
```
