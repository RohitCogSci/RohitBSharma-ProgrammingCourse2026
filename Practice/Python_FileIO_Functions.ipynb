```python
f= open("square.txt", "w")

for t in range(1, 11):
    sq = t*t
    f.write(str(t) + " " + str(sq) + "\n")
f.close()

print("squares updated to square.txt")
```

```python
with open("square.txt", "w") as f:
    for t in range(1,11):
        sq = t*t
        f.write(str(t) + " " + str(sq) + "\n")
print("file closed")
```
